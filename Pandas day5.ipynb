```python
import pandas as pd
```

```python
df_emp=pd.read_csv('pandayday1\emp1.csv')
df_emp
```

```python
df_Sal=pd.read_csv('pandayday1\salary.csv')
```

```python
df_emp.merge(df_Sal,left_on='eno',right_on='eno1',how='inner')
```

```python
df_left=df_emp.merge(df_Sal,left_on='eno',right_on='eno1',how='left')
df_left
```

```python
df_left['eno1'].isna()
```

```python
df_left[df_left['eno1'].isna()]
```

```python
df_left[df_left['eno1'].notna()]
```

```python
df_right=df_emp.merge(df_Sal,left_on='eno',right_on='eno1',how='right')
```

```python
df_right
```

```python
df_right[df_right['eno'].isna()]
```

```python
df_right[df_right['eno'].notna()]
```

```python
df_outer=df_emp.merge(df_Sal,left_on='eno',right_on='eno1',how='outer')
```

```python
df_outer
```

```python
df_outer[(df_outer['eno'].isna()) | (df_outer['eno1'].isna())]
```

```python
df_outer[(df_outer['eno'].notna()) & (df_outer['eno1'].notna())]
```

```python
df_outer.to_csv('outer_join_results.csv',index=False)
```
